```python
import numpy as np
import matplotlib.pyplot as plt

# Диапазон значений дискриминантной функции d(x)
d = np.linspace(-5, 5, 500)

# Рассматриваемые значения сигмы
sigmas = [0.5, 2.0]
colors = ['#1f77b4', '#d62728']  # синий и красный

plt.figure(figsize=(9, 6))

for sigma, color in zip(sigmas, colors):
    # В вашей постановке d+ - d- = 1, поэтому gamma = 1 / sigma^2
    gamma = 1 / (sigma**2)

    # Формула апостериорной вероятности при равных априорах P(+) = P(-) = 0.5
    # P(+|d) = 1 / (1 + exp(-gamma * d))
    P_plus = 1 / (1 + np.exp(-gamma * d))

    plt.plot(d, P_plus, label=f'σ = {sigma:.2f}, γ = {gamma:.2f}', color=color)

plt.xlabel('d(x)  —  скор (проекция на решающее направление)', fontsize=13)
plt.ylabel('P(+ | d)  —  апостериорная вероятность класса "+"', fontsize=13)
plt.title('Влияние внутреклассового разброса σ на форму логистической кривой', fontsize=14)
plt.legend(fontsize=12, loc='upper left')
plt.grid(True)
plt.show()
```
