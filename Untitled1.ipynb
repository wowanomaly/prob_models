```python
import numpy as np
```

```python
S = np.array([[0, -1, 0], [-1/2, 0, 1/2], [1/2, 1, 1/2]])
S_minus_one = np.linalg.inv(S)

print('Matrix S:')
print(S)

print('Matrix S^-1:')
print(S_minus_one)

D_standard = np.array([[0, 1, 0], [0, 0, 2], [0, 0, 0]])
D_star_standard = np.transpose(D_standard)

print('Matrix D in standard base:')
print(D_standard)

print('Matrix D* in standard base:')
print(D_star_standard)

D_new = np.matmul(np.matmul(S_minus_one, D_standard), S)
D_star_new = np.matmul(np.matmul(S_minus_one, D_star_standard), S)

print('Matrix D in new base:')
print(D_new)

print('Matrix D* in new base:')
print(D_star_new)
```
