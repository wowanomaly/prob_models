```python
import numpy as np

def pca(X, n_components=None):
    """
    Метод главных компонент (PCA).

    Параметры
    ---------
    X : np.ndarray, форма (n_samples, n_features)
        Исходные данные.
    n_components : int, optional
        Количество главных компонент для сохранения.
        Если None, сохраняются все компоненты.

    Возвращает
    ----------
    X_projected : np.ndarray, форма (n_samples, n_components)
        Проекция данных на первые n_components главных компонент.
    components : np.ndarray, форма (n_components, n_features)
        Главные компоненты (направления максимальной дисперсии) в виде строк.
    explained_variance : np.ndarray, форма (n_components,)
        Собственные значения ковариационной матрицы – дисперсия вдоль каждой компоненты.
    """
    # 1. Центрирование данных (вычитание среднего по каждому признаку)
    mean = np.mean(X, axis=0)
    X_centered = X - mean

    # 2. Вычисление ковариационной матрицы (несмещённая оценка)
    n_samples = X.shape[0]
    cov_matrix = (X_centered.T @ X_centered) / (n_samples - 1)

    # 3. Собственные значения и векторы ковариационной матрицы
    #    (используем eigh для симметричных матриц – быстрее и устойчивее)
    eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

    # 4. Сортировка по убыванию собственных значений
    sorted_indices = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[sorted_indices]
    eigenvectors = eigenvectors[:, sorted_indices]

    # 5. Выбор нужного числа компонент
    if n_components is None:
        n_components = X.shape[1]
    eigenvectors = eigenvectors[:, :n_components]
    eigenvalues = eigenvalues[:n_components]

    # 6. Проецирование данных на главные компоненты
    X_projected = X_centered @ eigenvectors

    # 7. Формирование выходных матриц:
    #    компоненты возвращаем как строки (удобно для интерпретации)
    components = eigenvectors.T

    return X_projected, components, eigenvalues
```

```python
# Генерируем синтетические данные
np.random.seed(42)
n_samples = 100
n_features = 5

# Создаём данные с корреляцией: первые два признака линейно зависимы с шумом
X = np.random.randn(n_samples, n_features)
X[:, 1] = 0.8 * X[:, 0] + 0.2 * np.random.randn(n_samples)  # зависимость от первого признака
X[:, 3] = -0.6 * X[:, 2] + 0.3 * np.random.randn(n_samples) # зависимость от третьего

# Применяем PCA, сохраняем все компоненты
X_proj, comp, ev = pca(X, n_components=None)

print("Форма исходных данных:", X.shape)
print("Форма проекции (все компоненты):", X_proj.shape)
print("Форма главных компонент (строки):", comp.shape)
print("Собственные значения (дисперсия вдоль компонент):", ev)

# Доля объяснённой дисперсии
explained_variance_ratio = ev / np.sum(ev)
print("\nДоля объяснённой дисперсии для каждой компоненты:")
for i, ratio in enumerate(explained_variance_ratio):
    print(f"  PC{i+1}: {ratio:.4f}")

# Кумулятивная объяснённая дисперсия
cumulative = np.cumsum(explained_variance_ratio)
print("\nКумулятивная доля объяснённой дисперсии:")
for i, cum in enumerate(cumulative):
    print(f"  PC{i+1}: {cum:.4f}")

# Выбираем первые две компоненты для визуализации (если нужно)
X_proj_2, comp_2, ev_2 = pca(X, n_components=2)
print("\n--- Проекция на 2 компоненты ---")
print("Форма проекции (2 компоненты):", X_proj_2.shape)

# Восстановление данных из двух компонент (с учётом центрирования)
mean = np.mean(X, axis=0)
X_reconstructed = X_proj_2 @ comp_2 + mean
print("Форма восстановленных данных (из 2 компонент):", X_reconstructed.shape)

# Среднеквадратичная ошибка восстановления (пример)
mse = np.mean((X - X_reconstructed)**2)
print(f"Среднеквадратичная ошибка восстановления (2 компоненты): {mse:.6f}")
```
