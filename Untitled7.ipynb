# Дискриминантный анализ: классификация фирм по уклонению от уплаты налогов

## Постановка задачи

Есть 2 обучающие выборки, соответствующие 2 фирмам:
- **Класс 1**: фирмы, уклоняющиеся от уплаты налогов
- **Класс 2**: фирмы, не уклоняющиеся от уплаты налогов

Необходимо определить, к какой совокупности следует отнести фирму с показателями:
$$x_0^{(1)} = 740, \quad x_0^{(2)} = 590$$

## 1. Вычисление основных параметров

### Формулы для вычислений:

**Средние векторы для каждой группы:**
$$a^{(j)} = \frac{1}{n_j} \sum_{i=1}^{n_j} X_{ji}, \quad j = 1, 2$$

**Ковариационные матрицы для каждой группы:**
$$\sigma^{(j)} = \frac{1}{n_j} \sum_{i=1}^{n_j} (X_{ji} - a^{(j)})(X_{ji} - a^{(j)})^T$$

**Пулированная (объединенная) ковариационная матрица:**
$$\sigma = \frac{1}{n_1 + n_2 - 2} \left(n_1 \sigma^{(1)} + n_2 \sigma^{(2)}\right)$$

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Загрузка данных
df = pd.read_csv('data.csv', sep=';')

# Разделяем на две группы по признаку "Тип выборки"
X1 = df[df['Тип выборки'] == 'уклоняющиеся'][['x1', 'x2']].values
X2 = df[df['Тип выборки'] == 'не уклоняющиеся'][['x1', 'x2']].values

n1, n2 = len(X1), len(X2)
print(f"Размеры выборок: n1={n1}, n2={n2}\n")

# 2. Вычисление средних векторов a(1), a(2)
a1 = np.mean(X1, axis=0)
a2 = np.mean(X2, axis=0)
print(f"a(1) = {a1}")
print(f"a(2) = {a2}\n")

# 3. Вычисление ковариационных матриц sigma(1), sigma(2)
diff1 = X1 - a1
sigma1 = (diff1.T @ diff1) / n1

diff2 = X2 - a2
sigma2 = (diff2.T @ diff2) / n2
print(f"sigma(1):\n{sigma1}\n")
print(f"sigma(2):\n{sigma2}\n")

# 4. Вычисление пулированной ковариационной матрицы sigma
sigma = (n1 * sigma1 + n2 * sigma2) / (n1 + n2 - 2)
print(f"Пул. ковариация sigma:\n{sigma}\n")

# 5. Вычисление обратной матрицы sigma^-1
sigma_inv = np.linalg.inv(sigma)
print(f"sigma^-1:\n{np.round(sigma_inv, 6)}\n")
```

## 2. Решающее правило дискриминантного анализа

Для классификации новой фирмы вычисляем:

**Вектор весов:**
$$w = \sigma^{-1}(a^{(1)} - a^{(2)})$$

**Дискриминантная функция:**
$$g(X_0) = \left[X_0 - \frac{1}{2}(a^{(1)} + a^{(2)})\right]^T \sigma^{-1} (a^{(1)} - a^{(2)})$$

**Решающее правило:**
$$\text{Если } g(X_0) \geq \ln\left(\frac{\pi_2}{\pi_1}\right) \rightarrow \text{класс 1}$$
$$\text{Иначе } \rightarrow \text{класс 2}$$

**Пояснение:** $\ln(\pi_2/\pi_1) = 0$, так как при отсутствии априорной информации о вероятностях классов принимаем $\pi_1 = \pi_2 = 0.5$, следовательно $\ln(0.5/0.5) = \ln(1) = 0$

```python
# 6. Вектор новой фирмы
X0 = np.array([740, 590])

# 7. Вычисление решающего дискриминантного выражения
diff_means = a1 - a2
mid_point = 0.5 * (a1 + a2)
w = sigma_inv @ diff_means  # sigma^-1 * (a(1) - a(2))
score = (X0 - mid_point) @ w  # скалярное произведение

print(f"Вектор w = sigma^-1 * (a(1) - a(2)): {w}")
print(f"Значение дискриминантной функции: {score:.4f}")

# 8. Пороговое значение ln(pi2/pi1)
threshold = np.log(0.5 / 0.5)
print(f"ln(pi2/pi1) = {threshold:.4f} (т.к. приняты равные априорные вероятности pi1=pi2=0.5)\n")

# 9. Применение решающего правила
if score >= threshold:
    class_pred = 1
    class_name = "уклоняющиеся от уплаты налогов (класс 1)"
else:
    class_pred = 2
    class_name = "не уклоняющиеся от уплаты налогов (класс 2)"

print(f"РЕШЕНИЕ: Фирма относится к классу {class_pred} ({class_name})\n")
```

## 3. Расстояние Махаланобиса

Расстояние Махаланобиса измеряет степень различия двух анализируемых совокупностей:

$$D^2 = (a^{(1)} - a^{(2)})^T \sigma^{-1} (a^{(1)} - a^{(2)})$$

$$D = \sqrt{D^2}$$

```python
# 10. Вычисление расстояния Махаланобиса между совокупностями
D_sq = diff_means @ w
D = np.sqrt(D_sq)
print(f"Расстояние Махаланобиса:")
print(f"D^2 = {D_sq:.4f}")
print(f"D   = {D:.4f}")
```

## 4. Визуализация результатов

На графике отображены:
- Фирмы, уклоняющиеся от налогов (класс 1) - синие кружки
- Фирмы, не уклоняющиеся от налогов (класс 2) - красные квадраты
- Классифицируемая фирма $X_0$ - черный крестик 'X'
- Центроиды групп - большие маркеры

```python
# Визуализация
plt.figure(figsize=(10, 8))

# Отображение фирм класса 1 (уклоняющиеся)
plt.scatter(X1[:, 0], X1[:, 1], c='blue', marker='o', s=100,
            label='Уклоняющиеся (класс 1)', alpha=0.6, edgecolors='black')

# Отображение фирм класса 2 (не уклоняющиеся)
plt.scatter(X2[:, 0], X2[:, 1], c='red', marker='s', s=100,
            label='Не уклоняющиеся (класс 2)', alpha=0.6, edgecolors='black')

# Отображение классифицируемой фирмы X0
plt.scatter(X0[0], X0[1], c='black', marker='X', s=200,
            label=f'Фирма X0 = {class_name}', edgecolors='black', linewidths=2)

# Отображение центроидов групп
plt.scatter(a1[0], a1[1], c='blue', marker='*', s=300,
            label=f'Центроид 1: ({a1[0]:.1f}, {a1[1]:.1f})',
            edgecolors='black', linewidths=2)
plt.scatter(a2[0], a2[1], c='red', marker='*', s=300,
            label=f'Центроид 2: ({a2[0]:.1f}, {a2[1]:.1f})',
            edgecolors='black', linewidths=2)

# Добавление подписей и заголовка
plt.xlabel('x1', fontsize=12)
plt.ylabel('x2', fontsize=12)
plt.title('Дискриминантный анализ: классификация фирм\n'
          f'X0 = [{X0[0]}, {X0[1]}] → {class_name}',
          fontsize=14, fontweight='bold')

plt.legend(loc='best', fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Дополнительная информация
print("\n" + "="*60)
print("ИТОГОВАЯ ИНФОРМАЦИЯ:")
print("="*60)
print(f"Классифицируемая фирма: X0 = [{X0[0]}, {X0[1]}]")
print(f"Значение дискриминантной функции: g(X0) = {score:.4f}")
print(f"Пороговое значение: ln(π2/π1) = {threshold:.4f}")
print(f"Расстояние Махаланобиса: D = {D:.4f} (D² = {D_sq:.4f})")
print(f"\nРЕШЕНИЕ: Фирма отнесена к классу {class_pred} - {class_name}")
print("="*60)
```
