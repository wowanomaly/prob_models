```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, laplace, cauchy
```

```python
x = np.linspace(-10, 10, 1000)

mu = 0
sigma = 1

normal_pdf = norm.pdf(x, mu, sigma)
laplace_pdf = laplace.pdf(x, mu, sigma)
cauchy_pdf = cauchy.pdf(x, mu, sigma)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
fig.suptitle('Сравнение распределений вероятности')

axes[0].plot(x, normal_pdf, 'b-')
axes[0].fill_between(x, normal_pdf, alpha=0.3, color='blue')
axes[0].set_title('Нормальное распределение')
axes[0].set_xlabel('x')
axes[0].grid(True, alpha=0.3)
axes[0].set_ylim(0, 0.45)

axes[1].plot(x, laplace_pdf, 'r-')
axes[1].fill_between(x, laplace_pdf, alpha=0.3, color='red')
axes[1].set_title('Распределение Лапласа')
axes[1].set_xlabel('x')
axes[1].grid(True, alpha=0.3)
axes[1].set_ylim(0, 0.55)

axes[2].plot(x, cauchy_pdf, 'g-')
axes[2].fill_between(x, cauchy_pdf, alpha=0.3, color='green')
axes[2].set_title('Распределение Коши')
axes[2].set_xlabel('x')
axes[2].grid(True, alpha=0.3)
axes[2].set_ylim(0, 0.35)

plt.tight_layout()
plt.show()

plt.figure(figsize=(15, 4))
plt.plot(x, normal_pdf, 'b-', linewidth=2.5, label='Нормальное')
plt.plot(x, laplace_pdf, 'r-', linewidth=2.5, label='Лапласа')
plt.plot(x, cauchy_pdf, 'g-', linewidth=2.5, label='Коши')
plt.xlabel('x')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
a = 0.5
b_line = 2.0
n_points = 100
x_data = np.linspace(0, 10, n_points)
y_line = a * x_data + b_line
```

```python
noise_normal = np.random.normal(mu, sigma, n_points)
noise_laplace = np.random.laplace(mu, sigma, n_points)
noise_cauchy = mu + sigma * np.random.standard_cauchy(n_points)
```

```python
y_normal = y_line + noise_normal
y_laplace = y_line + noise_laplace
y_cauchy = y_line + noise_cauchy
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

axes[0].scatter(x_data, y_normal, alpha=0.6)
axes[0].plot(x_data, y_line, 'r-', linewidth=2.5, label=f'y = {a}x + {b_line}')
axes[0].set_title('Нормальное распределение')
axes[0].set_xlabel('x')
axes[0].set_ylabel('y')
axes[0].grid(True, alpha=0.3)
axes[0].legend()
axes[0].set_ylim(-3, 8)

axes[1].scatter(x_data, y_laplace, alpha=0.6)
axes[1].plot(x_data, y_line, 'g-', linewidth=2.5, label=f'y = {a}x + {b_line}')
axes[1].set_title('Распределение Лапласа', fontsize=12)
axes[1].set_xlabel('x')
axes[1].set_ylabel('y')
axes[1].grid(True, alpha=0.3)
axes[1].legend()
axes[1].set_ylim(-3, 8)

axes[2].scatter(x_data, y_cauchy, alpha=0.6)
axes[2].plot(x_data, y_line, 'b-', linewidth=2.5, label=f'y = {a}x + {b_line}')
axes[2].set_title('Распределение Коши')
axes[2].set_xlabel('x')
axes[2].set_ylabel('y')
axes[2].grid(True, alpha=0.3)
axes[2].legend()
axes[2].set_ylim(-15, 20)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def get_cov_matrix(s1, s2, rho):
    return [[s1**2, rho * s1 * s2],
            [rho * s1 * s2, s2**2]]

def plot_all_cases():
    n_samples = 300
    mu_p = np.array([1, 1])
    mu_m = np.array([-1, -1])

    # Область для контуров
    x_range = np.linspace(-4, 4, 400)
    y_range = np.linspace(-4, 4, 400)
    X, Y = np.meshgrid(x_range, y_range)

    fig, axs = plt.subplots(2, 2, figsize=(12, 12))
    axs = axs.flatten()

    # ---------- СЛУЧАЙ 1 ----------
    # Σ+ = Σ- = I
    cov1 = get_cov_matrix(1, 1, 0)
    data_p1 = np.random.multivariate_normal(mu_p, cov1, n_samples)
    data_m1 = np.random.multivariate_normal(mu_m, cov1, n_samples)

    axs[0].scatter(data_p1[:, 0], data_p1[:, 1], alpha=0.4, s=10, c='blue', label='+')
    axs[0].scatter(data_m1[:, 0], data_m1[:, 1], alpha=0.4, s=10, c='red', label='-')
    axs[0].contour(X, Y, X + Y, levels=[0], colors='black', linewidths=2)
    axs[0].set_title("Случай 1: Σ = I\n$x_1 + x_2 = 0$")

    # ---------- СЛУЧАЙ 2 ----------
    # Σ = [[1, 1], [1, 2]]
    cov2 = get_cov_matrix(1, np.sqrt(2), np.sqrt(2)/2)
    data_p2 = np.random.multivariate_normal(mu_p, cov2, n_samples)
    data_m2 = np.random.multivariate_normal(mu_m, cov2, n_samples)

    axs[1].scatter(data_p2[:, 0], data_p2[:, 1], alpha=0.4, s=10, c='blue', label='+')
    axs[1].scatter(data_m2[:, 0], data_m2[:, 1], alpha=0.4, s=10, c='red', label='-')
    axs[1].contour(X, Y, X, levels=[0], colors='black', linewidths=2)   # x1 = 0
    axs[1].set_title("Случай 2: Σ = [[1,1],[1,2]]\n$x_1 = 0$")

    # ---------- СЛУЧАЙ 3 ----------
    # Σ+ = Σ- = [[1, ρ], [ρ, 1]] с ρ = 0.5 (граница x1 + x2 = 0)
    rho = 0.5
    cov3 = get_cov_matrix(1, 1, rho)
    data_p3 = np.random.multivariate_normal(mu_p, cov3, n_samples)
    data_m3 = np.random.multivariate_normal(mu_m, cov3, n_samples)

    axs[2].scatter(data_p3[:, 0], data_p3[:, 1], alpha=0.4, s=10, c='blue', label='+')
    axs[2].scatter(data_m3[:, 0], data_m3[:, 1], alpha=0.4, s=10, c='red', label='-')
    axs[2].contour(X, Y, X + Y, levels=[0], colors='black', linewidths=2)
    axs[2].set_title(f"Случай 3: Σ = [[1,{rho}],[{rho},1]]\n$x_1 + x_2 = 0$")

    # ---------- СЛУЧАЙ ρ и -ρ ----------
    # Σ+ = [[1, ρ], [ρ, 1]] , Σ- = [[1, -ρ], [-ρ, 1]]
    rho = 0.5
    cov_p = get_cov_matrix(1, 1, rho)
    cov_m = get_cov_matrix(1, 1, -rho)
    data_p4 = np.random.multivariate_normal(mu_p, cov_p, n_samples)
    data_m4 = np.random.multivariate_normal(mu_m, cov_m, n_samples)

    # Уравнение границы: ρ x1 x2 + a(x1+x2) + a^2 ρ = 0, здесь a=1
    Z = rho * X * Y + 1 * (X + Y) + 1**2 * rho

    axs[3].scatter(data_p4[:, 0], data_p4[:, 1], alpha=0.4, s=10, c='blue', label='+')
    axs[3].scatter(data_m4[:, 0], data_m4[:, 1], alpha=0.4, s=10, c='red', label='-')
    axs[3].contour(X, Y, Z, levels=[0], colors='black', linewidths=2)
    axs[3].set_title(f"Случай ρ и -ρ: ρ = {rho}\nГипербола: ρ x1 x2 + x1 + x2 + ρ = 0")

    # Оформление всех осей
    for ax in axs:
        ax.axhline(0, color='grey', lw=1, ls='--')
        ax.axvline(0, color='grey', lw=1, ls='--')
        ax.set_xlim([-4, 4])
        ax.set_ylim([-4, 4])
        ax.legend(loc='upper right')
        ax.grid()

    plt.tight_layout()
    plt.show()

plot_all_cases()
```
