```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------
# 1. Generate synthetic data
# ------------------------------
np.random.seed(42)  # for reproducibility

# Number of samples per class
n_samples = 1000

# Class 0: centered at (-2, -2)
mean0 = [-2, -2]
cov0 = [[1.5, 0.5], [0.5, 1.5]]
X0 = np.random.multivariate_normal(mean0, cov0, n_samples)
y0 = np.zeros(n_samples)  # label 0

# Class 1: centered at (2, 2)
mean1 = [2, 2]
cov1 = [[1.5, 0.5], [0.5, 1.5]]
X1 = np.random.multivariate_normal(mean1, cov1, n_samples)
y1 = np.ones(n_samples)   # label 1

# Combine and shuffle
X = np.vstack((X0, X1))
y = np.hstack((y0, y1))
indices = np.random.permutation(len(y))
X, y = X[indices], y[indices]

# ------------------------------
# 2. Perceptron implementation
# ------------------------------
class Perceptron:
    def __init__(self, learning_rate=0.01, max_epochs=1000):
        self.lr = learning_rate
        self.max_epochs = max_epochs
        self.weights = None
        self.bias = None

    def fit(self, X, y):
        n_samples, n_features = X.shape
        # Initialize weights and bias to small random values
        self.weights = np.random.randn(n_features) * 0.01
        self.bias = 0.0

        for epoch in range(self.max_epochs):
            errors = 0
            for xi, target in zip(X, y):
                # Linear combination
                linear_output = np.dot(xi, self.weights) + self.bias
                # Step activation (0 if <0, 1 if >=0)
                prediction = 1 if linear_output >= 0 else 0

                # Update if misclassified
                if prediction != target:
                    error = target - prediction
                    self.weights += self.lr * error * xi
                    self.bias   += self.lr * error
                    errors += 1

            if errors == 0:
                print(f"Converged after {epoch+1} epochs")
                break

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, 0)

# ------------------------------
# 3. Train the perceptron
# ------------------------------
perceptron = Perceptron(learning_rate=0.1, max_epochs=1000)
perceptron.fit(X, y)

# ------------------------------
# 4. Visualize results
# ------------------------------
# Create a mesh grid for plotting decision boundary
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))
grid = np.c_[xx.ravel(), yy.ravel()]
Z = perceptron.predict(grid).reshape(xx.shape)

# Plot
plt.figure(figsize=(10, 8))
plt.title('Binary Linear Classification with Perceptron')
plt.contourf(xx, yy, Z, alpha=0.5, cmap=plt.cm.RdYlBu)
plt.scatter(X[y==0, 0], X[y==0, 1], color='red', marker='o', label='Class 0', edgecolor='k')
plt.scatter(X[y==1, 0], X[y==1, 1], color='blue', marker='s', label='Class 1', edgecolor='k')

# Decision boundary line: w0*x + w1*y + b = 0  =>  y = -(w0*x + b)/w1
w0, w1 = perceptron.weights
b = perceptron.bias
x_vals = np.array([x_min, x_max])
y_vals = -(w0 * x_vals + b) / w1
plt.plot(x_vals, y_vals, 'g--', linewidth=3, label='Decision boundary')

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid()
plt.show()

# ------------------------------
# 5. Compute distances to decision boundary
# ------------------------------
# Perpendicular distance formula: |w0*x + w1*y + b| / ||w||
w_norm = np.sqrt(w0**2 + w1**2)
distances = np.abs(np.dot(X, perceptron.weights) + perceptron.bias) / w_norm

# Separate distances by true class
dist_class0 = distances[y == 0]
dist_class1 = distances[y == 1]

# Plot histogram of distances
plt.figure(figsize=(8, 5))
plt.title('Distribution of Distances from Points to the Separating Line')
plt.hist(dist_class0, bins=20, color='red', label='Class 0', edgecolor='black')
plt.xlabel('Distance to Decision Boundary')
plt.ylabel('Number of Points')
plt.legend()
plt.grid()
plt.show()

# Plot histogram of distances
plt.figure(figsize=(8, 5))
plt.title('Distribution of Distances from Points to the Separating Line')
plt.hist(dist_class1, bins=20, color='blue', label='Class 1', edgecolor='black')
plt.xlabel('Distance to Decision Boundary')
plt.ylabel('Number of Points')
plt.legend()
plt.grid()
plt.show()
```
