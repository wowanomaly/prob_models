```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import scipy
from sklearn.datasets import make_regression
from sklearn.preprocessing import StandardScaler
from ipywidgets import interact, interact_manual, Label
```

```python
#@interact(mean_x1=(-10, 10, 1), mean_x2=(-10, 10, 1), cov_11=(0, 10, 1), cov_12=(0, 10, 1), cov_22=(0, 10, 1))
def plot_md(mean_x1=(0.0), mean_x2=(0.0), cov_11=(1.0), cov_12=(1.0), cov_22=(1.0)):
  mean_vector = np.array([mean_x1, mean_x2])
  cov_matrix = [
        [cov_11, cov_12],
        [cov_12, cov_22]
      ]
  x, y = np.random.multivariate_normal(mean_vector, cov_matrix, 100).T
  plt.plot(x, y, '.')
  plt.axis('equal')
  plt.xlabel('$x_1$')
  plt.ylabel('$x_2$')
  plt.show()
```
